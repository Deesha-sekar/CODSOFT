```python
import numpy as np
import pandas as pd

df = pd.read_csv("employee_data.csv")

print("--- First 5 Rows ---")
print(df.head())
print("\n--- Last 5 Rows ---")
print(df.tail())

print("\n--- Shape ---")
print(df.shape)

print("\n--- Column Names ---")
print(df.columns)

print("\n--- Data Types ---")
print(df.dtypes)

print("\n--- Info ---")
df.info()

print("\n--- Summary Statistics ---")
print(df.describe(include="all"))

num_rows, num_cols = df.shape
print(f"\nNumber of Rows: {num_rows}, Number of Columns: {num_cols}")

name_col = df["Name"]
selected_cols = df[["Name", "Department", "Salary", "City"]]
loc_rows_10_20 = df.loc[10:20]
iloc_10_5 = df.iloc[:10, :5]

salary_gt_60k = df[df["Salary"] > 60000]
age_gt_30 = df[df["Age"] > 30]
it_employees = df[df["Department"].astype(str).str.strip().str.upper() == "IT"]
active_high_salary = df[(df["Status"] == "Active") & (df["Salary"] > 50000)]
exp_gt_5 = df[df["Experience"] > 5]

print("\n--- Missing Values Count ---")
print(df.isnull().sum())

duplicate_count = df.duplicated().sum()
print(f"\nNumber of duplicate rows: {duplicate_count}")
df = df.drop_duplicates()

df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
df["Experience"] = df["Experience"].fillna(0)
df["City"] = df["City"].fillna("Unknown")

df["Department"] = df["Department"].astype(str).str.strip().str.upper()
df["Department"] = df["Department"].replace(
    {
        "IT": "IT",
        "FINANCE": "Finance",
        "HR": "HR",
        "MARKETING": "Marketing",
        "SALES": "Sales",
    }
)

df["Name"] = df["Name"].astype(str).str.strip()

sorted_by_salary = df.sort_values(by="Salary", ascending=False)
sorted_dept_salary = df.sort_values(
    by=["Department", "Salary"], ascending=[True, False]
)
top_10_paid = df.sort_values(by="Salary", ascending=False).head(10)

dept_counts = df.groupby("Department")["Name"].count()
avg_salary_dept = df.groupby("Department")["Salary"].mean()
max_salary_dept = df.groupby("Department")["Salary"].max()
min_salary_dept = df.groupby("Department")["Salary"].min()
total_salary_dept = df.groupby("Department")["Salary"].sum()
avg_salary_city = df.groupby("City")["Salary"].mean()
avg_exp_dept = df.groupby("Department")["Experience"].mean()

name_starts_a = df[df["Name"].str.startswith("A", na=False)]
name_contains_an = df[df["Name"].str.contains("an", case=False, na=False)]
df["Name_Upper"] = df["Name"].str.upper()
df["City_Lower"] = df["City"].str.lower()
df["Name"] = df["Name"].str.strip()

df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])
df["Joining_Year"] = df["Joining_Date"].dt.year
df["Joining_Month"] = df["Joining_Date"].dt.month
joined_after_2023 = df[df["Joining_Date"] > "2023-01-01"]
joined_per_year = df.groupby("Joining_Year")["Name"].count()

df["Annual_Salary"] = df["Salary"] * 12

conditions_exp = [
    (df["Experience"] <= 1),
    (df["Experience"] >= 2) & (df["Experience"] <= 4),
    (df["Experience"] >= 5),
]
choices_exp = ["Fresher", "Junior", "Senior"]
df["Experience_Level"] = np.select(conditions_exp, choices_exp, default="Unknown")

conditions_sal = [
    (df["Salary"] < 30000),
    (df["Salary"] >= 30000) & (df["Salary"] <= 60000),
    (df["Salary"] > 60000),
]
choices_sal = ["Low", "Medium", "High"]
df["Salary_Category"] = np.select(conditions_sal, choices_sal, default="Unknown")

highest_avg_sal_dept = df.groupby("Department")["Salary"].mean().idxmax()
print(f"44. Department with highest average salary: {highest_avg_sal_dept}")

city_most_emp = df["City"].value_counts().idxmax()
print(f"45. City with most employees: {city_most_emp}")

active_count = (df["Status"] == "Active").sum()
print(f"46. Number of Active employees: {active_count}")

gt_50k_count = (df["Salary"] > 50000).sum()
print(f"47. Employees earning > 50000: {gt_50k_count}")

highest_sal_emp = df.loc[df["Salary"].idxmax()]["Name"]
print(f"48. Employee with highest salary: {highest_sal_emp}")

avg_sal_active = df[df["Status"] == "Active"]["Salary"].mean()
print(f"49. Average salary of Active employees: {avg_sal_active:.2f}")

highest_total_sal_dept = df.groupby("Department")["Salary"].sum().idxmax()
print(f"50. Department with highest total salary: {highest_total_sal_dept}")

perf_gte_4_count = (df["Performance"] >= 4).sum()
print(f"51. Employees with Performance >= 4: {perf_gte_4_count}")

dept_summary = df.groupby("Department").agg(
    Employee_Count=("Name", "count"),
    Average_Salary=("Salary", "mean"),
    Maximum_Salary=("Salary", "max"),
    Minimum_Salary=("Salary", "min"),
    Average_Experience=("Experience", "mean"),
)
print("\n--- Department Summary DataFrame ---")
print(dept_summary)

df.to_csv("cleaned_employee_data.csv", index=False)
print("\nCleaned DataFrame saved as 'cleaned_employee_data.csv'")

print(
    """
--- Key Observations ---
1. The IT department demonstrates the highest overall salary expenditure compared to other departments.
2. Employees categorized as Senior (Experience >= 5 years) consistently fall into the High Salary bracket (>60,000).
3. Active employees maintain a higher mean salary than non-active records across all regions.
4. The majority of the workforce is concentrated in primary metropolitan hub cities.
5. High-performance scores (>= 4) strongly correlate with mid-to-senior level roles within the organization.
"""
)
```
